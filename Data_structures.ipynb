```python
#Data structures
#Linked List
class Node:
    def __init__(self, data):
        self.data = data 
        self.next = None
        
class LinkedList:
    def __init__(self):
        self.head = None
    def insertion_at_head(self, n):
        new_node = Node(n)
        new_node.next = self.head
        self.head = new_node
    def insertion(self, k, n):
        if k == 0:
            self.insertion_at_head(n)
            return
        newNode = Node(n)
        curr = self.head
        for i in range(k - 1):
            if curr is None:
                print("out of bounds")
                curr = curr.next
        if curr is None:
            print("Index out of bounds")
        newNode.next = curr.next
        curr.next = newNode
    def deletion(self, k):
        if self.head == None:
            print("Linked list already empty")
        
        elif k == 0:
            self.head = self.head.next
            return
        else:
            curr = self.head
            for i in range(k-1):
                if curr is None or curr.next is None:
                    print("index out of bounds")
                    return
                curr = curr.next
            if curr.next is None:
                print("index out of bounds")
            curr.next = curr.next.next
    def display(self):
        print("Your current linked list: \n")
        curr = self.head
        while curr != None:
            print(curr.data, end = " ")
            curr = curr.next

my_list = LinkedList()

while True:
    print("choice 1: Head insertion | choice 2: insertion at k | choice 3: deletion at k | choice 4: display\n")
    try:
        choice = int(input("Enter your choice: "))
    except ValueError:
        print("Please enter a valid number.\n")
        continue
    if choice == 1:
        n = int(input("enter the number for head: "))
        my_list.insertion_at_head(n)
    elif choice == 2:
        n = int(input("enter number for insertion "))
        k = int(input("enter the index: "))
        my_list.insertion(k,n)
    elif choice == 3:
        n = int(input("Enter the index"))
        my_list.deletion(n)
    elif choice == 4:
        my_list.display()
    else:
        print("Invalid choice")        
```

```python
#Array implementation of stacks:
class stacks:
    def __init__(self):
        self.st = []
    def push(self, x):
        self.st.append(x)
    def pop(self):
        if len(self.st) == 0:
            print("Stack is empty")
            return -1
        else:
            x = self.st[-1]
            self.st.pop()
            return x
    def peek(self):
        if len(self.st) == 0:
            return -1
        else:
            return self.st[-1]
    def size(self):
        return len(self.st)
```

```python
#Linked list implementation of stack
class stack:
    def __init__(self):
        self.top = None
        self.length = 0
    def push(self,x):
        self.length += 1
        if self.top == None:
            self.top = Node(x)
            return
        else:
            newNode = Node(x)
            newNode.next = self.top
            self.top = newNode
    def pop(self):
        self.length -= 1
        if self.top == None:
            return -1
        x = self.top.data
        self.top = self.top.next
        return x
    def get_top(self):
        return self.top.data
    def get_length(self):
        return self.length
```
