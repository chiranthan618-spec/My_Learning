```python
%matplotlib inline
import matplotlib.pyplot as plt

years = [1900,1950,2000,2050,3000]
gdp = [300,375, 400,275,250]

plt.plot(years, gdp, color= "green", marker = "o", linestyle= "solid")

plt.title("Normal GDP")

plt.ylabel('Billions of $')
plt.show()
```

```python
#Bar chart:
movies = ["Avengers", "Kantara", "Inception","Interstellar", "Coco"]
num_oscars = [2,3,4,5,6]

plt.bar(range(len(movies)), num_oscars)

plt.title("My fav movies")

plt.ylabel("Academy awards")

plt.xticks(range(len(movies)), movies)

plt.show
```

```python
#Bar chart for plotting histograms
from collections import Counter
grades = [83,95,99,100,71,56, 45,100,91,83]

histogram = Counter(min(grades//10*10, 90) for grade in grades)

plt.bar([x+5 for x in histogram.keys()], histogram.values(), 10, edgecolor = (0,0,0))

plt.axis([-5,100,0,5])
```
