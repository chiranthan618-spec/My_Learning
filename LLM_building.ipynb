```python
#importing the file from github using urllib library
import urllib.request
url = ("https://raw.githubusercontent.com/rasbt/"
        "LLMs-from-scratch/main/ch02/01_main-chapter-code/"
        "the-verdict.txt")

file_path = "the-verdict.txt"

urllib.request.urlretrieve(url, file_path)

with open("the-verdict.txt", "r", encoding= "utf-8") as f:
    raw_text = f.read()

print("Total number of characters: ", len(raw_text))
print(raw_text[:99])
```

```python
#Building a simple tokenizer
import re
text = "Hello, world. This is a -- test."
result = re.split(r'([,.:;?_!"()\']|--|\s)', text)
result = [item.strip() for item in result if item.strip()]
print(result)
```

```python
#applying our  tokenizer model on the entire edith file we imported in 1

# This regex matches either a sequence of word characters (\w+) OR any single punctuation/special character
token_pattern = r'\w+|[ ,.:;?_!"()\'\-\[\]]'

# Use findall to extract every individual token
preprocessed = re.findall(token_pattern, raw_text)
preprocessed = [item.strip() for item in preprocessed if item.strip()]
print(len(preprocessed))
print(preprocessed[:30])
```

```python
#converting tokens into token id's
#creating a list of unique tokens and sort them alphabetically
all_words = sorted(set(preprocessed))
vocab_size = len(all_words)
print(vocab_size)

vocab = {token: integer for integer,token in enumerate(all_words)}
for i, item in enumerate(vocab.items()):
    print(item)
    if i >= 50:
        break
```
